```python
def optimize_contrast(folder_path, display, save):
    import skimage
    from skimage import io, exposure
    import matplotlib.pyplot as plt
    import os
    import numpy as np

    """Optimize the contrast of images in a folder and its subfolders, and display the results.

    Parameters
    ----------
    folder_path : str
        Path to the main folder containing the images and subfolders.
    display : bool
        Whether to display the original and optimized images.
    save : bool
        Whether to save the optimized images.

    Returns
    -------
    None
    """
    # Walk through the main folder and its subfolders
    for root, _, files in os.walk(folder_path):
        # Get a list of all image files in the current folder
        image_files = [file for file in files if file.endswith('.tiff') or file.endswith('.tif')]

        # Iterate over each image file
        for image_file in image_files:
            # Construct the full path to the image file
            image_path = os.path.join(root, image_file)

            # Read the image
            img = skimage.io.imread(image_path)

            # Normalize the image to be between -1 and 1
            img = img / np.max(np.abs(img))

            # Optimize the contrast of the image
            img_optimized = exposure.equalize_adapthist(img)

            # Specify the file path of the optimized image
            optimized_image_path = os.path.join(root, image_file)

            if save:
                # Save the optimized image
                skimage.io.imsave(optimized_image_path, img_optimized)

            if display:
                # Display the original image and the optimized image side by side
                fig, axes = plt.subplots(1, 2, figsize=(10, 5))
                axes[0].imshow(img, cmap='gray')
                axes[0].set_title('Original Image')
                axes[0].axis('off')
                axes[1].imshow(img_optimized, cmap='gray')
                axes[1].set_title('Optimized Image')
                axes[1].axis('off')
                plt.show()

# Example usage
optimize_contrast('/Users/allisonpickle/Desktop/DIV21-25_DEC2024/250405_151423_Experiment1/250405_151423_Plate 1/Normal Media', display=False, save=True)
```

```python
def Threshold_Test(folder_path):
    import os
    from skimage import io
    from skimage.filters import try_all_threshold
    import matplotlib.pyplot as plt
    """Test threshold options for images in a folder

    Parameters
    ----------
    folder_path : str
        Path to the folder containing the images

    Returns
    -------
    None
    """
    # Get a list of all image files in the folder
    image_files = [file for file in os.listdir(folder_path) if file.endswith('.png') or file.endswith('.tif')]

    # Iterate over each image file
    for image_file in image_files:
        # Construct the full path to the image file
        image_path = os.path.join(folder_path, image_file)

        # Read the image
        img = io.imread(image_path)
        fig, ax = try_all_threshold(img, figsize=(20, 18), verbose=False)
        plt.title(image_file)  # Add the image file name as the title
        plt.show()
Threshold_Test(folder_path)
```

```python
def Apply_mean_threshold(folder_path): #Auto apply to only astrocytes and neurons
    import os
    from skimage import io, filters
    """Apply mean threshold filter to images in a folder

    Parameters
    ----------
    folder_path : str
        Path to the folder containing the images

    Returns
    -------
    None
    """
    # Create a subfolder for the thresholded images
    result_folder = os.path.join(folder_path, "Thresholded")
    os.makedirs(result_folder, exist_ok=True)

    # Get a list of all image files in the folder
    image_files = [file for file in os.listdir(folder_path) if file.endswith('.tif') or file.endswith('.tiff')]

    # Iterate over each image file
    for image_file in image_files:
        # Construct the full path to the image file
        image_path = os.path.join(folder_path, image_file)

        # Read the image
        img = io.imread(image_path)

        # Apply mean threshold filter
        threshold_value = filters.threshold_mean(img)
        binary_image = img > threshold_value

        # Save the thresholded image to the result folder
        result_image_path = os.path.join(result_folder, image_file)
        io.imsave(result_image_path, binary_image.astype('uint8') * 255)

Apply_mean_threshold(folder_path)
```

```python
def Remove_Islands(folder_path, disk_radius, display_results=False, save_results=True):
    from skimage.morphology import disk
    from skimage import morphology
    import os
    from skimage import io
    import matplotlib.pyplot as plt

    """
    Process all TIFF images in a folder by applying a white tophat transformation to remove islands.

    Parameters
    ----------
    folder_path : str
        Path to the folder containing the thresholded TIFF images.
    disk_radius : int
        Disk radius for the white tophat transformation.
    display_results : bool, optional
        Whether to display the processed images. Default is False.
    save_results : bool, optional
        Whether to save the processed images. Default is True.

    Returns
    -------
    None
    """
    # Get a list of all TIFF image files in the folder
    image_files = [file for file in os.listdir(folder_path) if file.endswith('.tiff') or file.endswith('.tif')]

    if not image_files:
        print("No TIFF files found in the folder.")
        return

    # Iterate over each image file
    for image_file in image_files:
        # Construct the full path to the image file
        image_path = os.path.join(folder_path, image_file)

        # Read the image file into a NumPy array
        image = io.imread(image_path)

        # Apply the white tophat transformation
        footprint = disk(disk_radius)
        res = morphology.white_tophat(image, footprint)

        # Subtract the result from the original image to remove islands
        processed_image = image - res

        # Save the processed image (overwrite the original file)
        if save_results:
            io.imsave(image_path, processed_image)

        # Display the processed image
        if display_results:
            plt.figure(figsize=(10, 10))
            plt.imshow(processed_image, cmap='gray')
            plt.title(f"Processed: {image_file}")
            plt.axis('off')
            plt.show()

Remove_Islands(folder_path, disk_radius=2, display_results=False, save_results=True)
```

```python
def Measure_Surface_Area_Coverage(folder_path, display_results=False, save_results=True):
    import cv2
    import numpy as np
    import os
    import pandas as pd
    from openpyxl import Workbook
    from openpyxl.utils.dataframe import dataframe_to_rows
    from openpyxl import load_workbook
    import matplotlib.pyplot as plt

    """
    Measure surface area coverage for thresholded images and save the results to an Excel sheet.

    Parameters
    ----------
    folder_path : str
        Path to the folder containing the thresholded image files.
    display_results : bool, optional
        Whether to display the images. Default is False.
    save_results : bool, optional
        Whether to save the results. Default is True.

    Returns
    -------
    None
    """
    # Create the output file path
    output_file = os.path.join(folder_path, "Surface_Area_Coverage.xlsx")

    if save_results:
        # Check if the Excel file already exists
        if os.path.exists(output_file):
            # Delete the existing file
            os.remove(output_file)
        
        # Create a new workbook
        workbook = Workbook()
        workbook.save(output_file)
        
        # Create a new folder to save the contour overlay images
        contour_folder = os.path.join(folder_path, "Surface Area Coverage Contours")
        os.makedirs(contour_folder, exist_ok=True)

    # Initialize the dictionary to store white area percentages
    white_area_percentages = {}

    # Iterate over each file in the folder
    for filename in os.listdir(folder_path):
        if filename.endswith(".tiff") or filename.endswith(".tif"):
            # Construct the full path to the image file
            image_path = os.path.join(folder_path, filename)

            # Read the image as a grayscale image
            image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

            # Count the number of white pixels in the binary mask
            white_pixels = np.count_nonzero(image)

            # Calculate the total number of pixels in the image
            total_pixels = image.shape[0] * image.shape[1]

            # Calculate the percentage of white area
            white_area_percentage = (white_pixels / total_pixels) * 100

            # Append the data to the dictionary using the original filename
            white_area_percentages[filename] = white_area_percentage

            # Plot the image with the white area outlined in purple
            image_rgb = cv2.cvtColor(image, cv2.COLOR_GRAY2RGB)
            contours, _ = cv2.findContours(image, cv2.RETR_LIST, cv2.CHAIN_APPROX_SIMPLE)
            cv2.drawContours(image_rgb, contours, -1, (128, 0, 128), 2)
            
            if display_results:
                plt.figure(figsize=(10, 10))
                plt.imshow(image_rgb)
                plt.title(f"{filename}")
                plt.axis('off')
                plt.show()
            
            if save_results:
                # Save the contour overlay image to the contour folder
                contour_image_path = os.path.join(contour_folder, f"{filename.split('.')[0]}.png")
                plt.imsave(contour_image_path, image_rgb)
                plt.close()

    if save_results:
        # Create a DataFrame from the white area percentages
        df = pd.DataFrame(list(white_area_percentages.items()), columns=["Image File", "White Area Percentage"])

        # Load the existing workbook
        workbook = load_workbook(output_file)

        # Select the active sheet
        sheet = workbook.active

        # Write the DataFrame to the sheet
        rows = dataframe_to_rows(df, index=False, header=True)
        for row in rows:
            sheet.append(row)

        # Save the workbook to the specified output file
        workbook.save(output_file)
        print(f"Results saved to {output_file}")

Measure_Surface_Area_Coverage(folder_path, display_results=False, save_results=True)
```

```python
def Apply_Watershed(folder_path, minimum_distance, display=True, save=True):
    """
    Apply the watershed algorithm to a folder of thresholded images, save the results to a "watershed" folder,
    and save the contour counts to an Excel file.

    Parameters
    ----------
    folder_path : str
        Path to the folder containing the thresholded images.
    minimum_distance : int
        Minimum distance between peaks for the watershed algorithm.
    display : bool, optional
        Whether to display the processed images. Default is True.
    save : bool, optional
        Whether to save the processed images and results. Default is True.

    Returns
    -------
    None
    """
    import os
    import cv2
    import numpy as np
    from skimage.feature import peak_local_max
    from skimage.segmentation import watershed
    from scipy import ndimage
    import matplotlib.pyplot as plt
    import pandas as pd

    # Get a list of all image files in the folder
    image_files = [file for file in os.listdir(folder_path) if file.endswith('.tiff') or file.endswith('.tif')]

    if not image_files:
        print("No TIFF files found in the folder.")
        return

    # Create a new folder for the watershed results
    watershed_folder = os.path.join(folder_path, "watershed")
    os.makedirs(watershed_folder, exist_ok=True)

    # Initialize a DataFrame to store contour counts
    results = pd.DataFrame(columns=["Image Name", "Contour Count"])

    # Iterate over each image file
    for image_file in image_files:
        # Construct the full path to the image file
        image_path = os.path.join(folder_path, image_file)

        # Read the thresholded image
        thresh = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

        # Compute the Euclidean distance transform
        D = ndimage.distance_transform_edt(thresh)

        # Find local maxima in the distance map
        coords = peak_local_max(D, min_distance=minimum_distance, labels=thresh)

        # Create a binary mask for the local maxima
        local_max = np.zeros_like(thresh, dtype=bool)
        for coord in coords:
            local_max[tuple(coord)] = True

        # Perform connected component analysis on the local maxima
        markers = ndimage.label(local_max, structure=np.ones((3, 3)))[0]

        # Apply the watershed algorithm
        labels = watershed(-D, markers, mask=thresh)

        # Count the number of unique labels (excluding background)
        contour_count = len(np.unique(labels)) - 1

        # Append the result to the DataFrame
        results = pd.concat([results, pd.DataFrame({"Image Name": [image_file], "Contour Count": [contour_count]})], ignore_index=True)

        # Create an output image to visualize the results
        output_image = cv2.cvtColor(thresh, cv2.COLOR_GRAY2BGR)
        for label in np.unique(labels):
            if label == 0:
                continue
            mask = np.zeros(thresh.shape, dtype="uint8")
            mask[labels == label] = 255
            contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
            cv2.drawContours(output_image, contours, -1, (0, 255, 0), 2)

        # Display the results
        if display:
            plt.figure(figsize=(10, 10))
            plt.imshow(cv2.cvtColor(output_image, cv2.COLOR_BGR2RGB))
            plt.title(f"Watershed Result: {image_file}")
            plt.axis('off')
            plt.show()

        # Save the results
        if save:
            # Save the output image to the "watershed" folder
            output_path = os.path.join(watershed_folder, f"{image_file}")
            cv2.imwrite(output_path, output_image)

    # Save the contour counts to an Excel sheet
    if save:
        excel_path = os.path.join(folder_path, "Contour_Counts.xlsx")
        results.to_excel(excel_path, index=False)


folder_path = '/Users/allisonpickle/Desktop/DIV21-25_DEC2024/250405_151423_Experiment1/250405_151423_Plate 1/Tri/Microglia/Thresholded'
Apply_Watershed(folder_path, minimum_distance=50, display=False, save=True)
```
