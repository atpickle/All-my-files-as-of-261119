```python
import pandas as pd
import os

def collapse_duplicate_rows(input_file, output_file=None, column='D'):
    """
    Collapse rows with duplicate values in a specified column.
    
    Parameters:
    -----------
    input_file : str
        Path to input Excel file
    output_file : str, optional
        Path to output Excel file. If None, creates a new file with '_collapsed' suffix
    column : str, default='D'
        Column letter or name to check for duplicates
    
    Returns:
    --------
    pandas.DataFrame
        The collapsed DataFrame
    """
    
    # Read the Excel file
    df = pd.read_excel(input_file)
    
    # If column is a letter (like 'D'), convert to column index
    if len(column) == 1 and column.isalpha():
        col_index = ord(column.upper()) - ord('A')
        col_name = df.columns[col_index]
    else:
        col_name = column
    
    print(f"Original shape: {df.shape}")
    print(f"Checking for duplicates in column: {col_name}")
    
    # Keep only the first occurrence of each value in the specified column
    df_collapsed = df.drop_duplicates(subset=[col_name], keep='first')
    
    print(f"Collapsed shape: {df_collapsed.shape}")
    print(f"Removed {df.shape[0] - df_collapsed.shape[0]} duplicate rows")
    
    # Create output filename if not provided
    if output_file is None:
        base, ext = os.path.splitext(input_file)
        output_file = f"{base}_collapsed{ext}"
    
    # Save to Excel
    df_collapsed.to_excel(output_file, index=False)
    print(f"\nSaved collapsed data to: {output_file}")
    
    return df_collapsed

# Example usage:
# Set your input file path here
input_file ='/Users/allisonpickle/Desktop/Sadi_Data/Lig data_raw_reform_225C.xlsx'

# Option 1: Auto-generate output filename (adds '_collapsed' to original name)
df_result = collapse_duplicate_rows(input_file, column='D')

```

```python
import pandas as pd

def append_columns_to_start(csv_file):
    """
    Reads a CSV file, appends the contents of columns at positions 5, 6, 7, and 8 (F, G, H, I)
    to the end of columns at positions 0, 1, 2, and 3 (A, B, C, D) respectively,
    and skips the first header row.

    Parameters:
    -----------
    csv_file : str
        Path to the input CSV file.

    Returns:
    --------
    pandas.DataFrame
        The modified DataFrame.
    """
    # Read the CSV file
    df = pd.read_csv(csv_file, skiprows=1)  # Skip the first header row

    # Get column names
    cols = df.columns.tolist()
    
    # Check if there are at least 9 columns
    if len(cols) < 9:
        raise ValueError(f"CSV file needs at least 9 columns. Found {len(cols)} columns: {cols}")

    # Append the contents of columns at index 5, 6, 7, 8 to columns at index 0, 1, 2, 3
    df.iloc[:, 0] = df.iloc[:, 0].astype(str) + df.iloc[:, 5].astype(str)
    df.iloc[:, 1] = df.iloc[:, 1].astype(str) + df.iloc[:, 6].astype(str)
    df.iloc[:, 2] = df.iloc[:, 2].astype(str) + df.iloc[:, 7].astype(str)
    df.iloc[:, 3] = df.iloc[:, 3].astype(str) + df.iloc[:, 8].astype(str)

    # Drop the original columns at positions 5, 6, 7, 8
    df = df.drop(columns=[cols[5], cols[6], cols[7], cols[8]])

    return df

# Example usage:
csv_file_path = '/Users/allisonpickle/Desktop/Sadi_Data/Crack Width/Crack_Width_225C.csv'
modified_df = append_columns_to_start(csv_file_path)
print(modified_df)

# Optional: Save to a new CSV file
# modified_df.to_csv('/Users/allisonpickle/Desktop/Sadi_Data/Crack Width/Crack_Width_225C_modified.csv', index=False)
```

```python
import pandas as pd

def append_columns_to_start(csv_file):
    """
    Reads a CSV file, appends the contents of columns F, G, H, and I to the end of columns A, B, C, and D respectively,
    and skips the first header row.

    Parameters:
    -----------
    csv_file : str
        Path to the input CSV file.

    Returns:
    --------
    pandas.DataFrame
        The modified DataFrame.
    """
    # Read the CSV file
    df = pd.read_csv(csv_file, skiprows=1)  # Skip the first header row

    # Ensure the columns F, G, H, and I exist
    required_columns = ['F', 'G', 'H', 'I']
    if not all(col in df.columns for col in required_columns):
        raise ValueError(f"One or more required columns {required_columns} are missing in the CSV file.")

    # Append the contents of columns F, G, H, and I to A, B, C, and D respectively
    df['A'] = df['A'].astype(str) + df['F'].astype(str)
    df['B'] = df['B'].astype(str) + df['G'].astype(str)
    df['C'] = df['C'].astype(str) + df['H'].astype(str)
    df['D'] = df['D'].astype(str) + df['I'].astype(str)

    # Drop the original columns F, G, H, and I
    df = df.drop(columns=required_columns)

    return df

# Example usage:
csv_file_path = '/Users/allisonpickle/Desktop/Sadi_Data/Crack Width/Crack_Width_225C.csv'
modified_df = append_columns_to_start(csv_file_path)
print(modified_df)
```
